# Reproduce: Five-Dimensional Agent Reliability

All bundled observations are synthetic demonstration data.

```python
from pathlib import Path
import json, sys
import pandas as pd
repo=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd()
sys.path.insert(0,str(repo/'src'))
from fdar import evaluate_runs
rows=[json.loads(x) for x in (repo/'data/synthetic_reference_runs.jsonl').read_text().splitlines()]
df=pd.DataFrame(rows)
scored=evaluate_runs(df)
scored.head()
```

```python
summary=scored.groupby('architecture')[['task_fidelity','recovery','safety','efficiency','observability','reliability_score']].mean(numeric_only=True)
summary
```

`recovery` is unobserved (NaN) for no-perturbation runs and is not treated as perfect recovery.
