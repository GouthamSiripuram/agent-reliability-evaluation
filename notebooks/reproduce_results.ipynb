# Reproduce: Five-Dimensional Agent Reliability

All bundled observations are synthetic demonstration data.

```python
from pathlib import Path
import json, sys
import pandas as pd
repo=Path.cwd().parent if Path.cwd().name=='notebooks' else Path.cwd()
sys.path.insert(0,str(repo/'src'))
from fdar import evaluate_runs
rows=[json.loads(x) for x in (repo/'data/synthetic_reference_runs.jsonl').read_text().splitlines()]
df=pd.DataFrame(rows)
scored=evaluate_runs(df)
scored.head()
```

```python
summary=scored.groupby(['architecture','observed_dimension_count'])[['task_fidelity','recovery','safety','efficiency','observability','reliability_score']].mean(numeric_only=True)
summary
```

Recovery is unobserved (NaN) when no failure was attempted. Compare summaries only within the same observation pattern. These are synthetic reference profiles; consistency inputs were assigned, not measured from repeated real-agent runs.
